## 1. Setup and quick recap

Import NumPy using the standard alias `np`.


```python
import numpy as np

marks = np.array([55, 72, 88, 64, 91])
print("Marks:", marks)
print("Shape:", marks.shape)
print("Number of dimensions:", marks.ndim)
```

## 2. Arithmetic operations on arrays

NumPy arithmetic is usually performed element by element when arrays have compatible shapes.


```python
a = np.array([10, 20, 30])
b = np.array([1, 2, 3])

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Floor division:", a // b)
print("Remainder:", a % b)
print("Power:", a ** 2)
```

### Scalar operations

A scalar is a single value. NumPy can apply it to each compatible array element.


```python
prices = np.array([100, 200, 300])

print("Add 10:", prices + 10)
print("Discount of 10%:", prices * 0.90)
print("Double each value:", prices * 2)
```

### Important: element-wise multiplication vs matrix multiplication

`a * b` multiplies corresponding elements. Matrix multiplication is a different operation and will be covered in the linear algebra lesson.


## 3. Universal functions (ufuncs)

NumPy provides functions that operate efficiently element by element across arrays.


```python
values = np.array([1, 4, 9, 16])

print("Square root:", np.sqrt(values))
print("Square:", np.square(values))
print("Absolute value:", np.abs(np.array([-3, -1, 0, 2])))
print("Exponential:", np.exp(np.array([0, 1])))
print("Natural log:", np.log(np.array([1, np.e, np.e ** 2])))
print("Rounded:", np.round(np.array([2.34, 5.67]), 1))
print("Floor:", np.floor(np.array([2.9, 5.1])))
print("Ceiling:", np.ceil(np.array([2.1, 5.1])))
```

Note: logarithms and square roots have mathematical domains. Some invalid inputs can produce `nan` or warnings, depending on the operation.


## 4. Comparisons and Boolean arrays

Comparisons return Boolean values: `True` or `False`.


```python
scores = np.array([45, 67, 82, 39, 91])

print("Scores > 60:", scores > 60)
print("Scores >= 50:", scores >= 50)
print("Scores == 45:", scores == 45)
print("Scores != 39:", scores != 39)
```

### Boolean indexing (filtering)

A Boolean mask can select only the elements whose condition is `True`.


```python
scores = np.array([45, 67, 82, 39, 91])

passed_filter = scores >= 50
passed_scores = scores[passed_filter]

print("Mask:", passed_filter)
print("Passing scores:", passed_scores)
```

### Combine multiple conditions

Use `&` for element-wise AND and `|` for element-wise OR. Put each condition inside parentheses. Do not use Python's `and` or `or` for element-wise array comparisons.


```python
scores = np.array([45, 67, 82, 39, 91])

# Select scores from 50 through 80, inclusive
selected = scores[(scores >= 50) & (scores <= 80)]
print("Scores between 50 and 80:", selected)

# Select scores below 40 or above 90
extreme = scores[(scores < 40) | (scores > 90)]
print("Very low or very high scores:", extreme)
```

## 5. Conditional selection with `np.where()`

`np.where(condition, value_if_true, value_if_false)` chooses values based on a condition.


```python
scores = np.array([45, 67, 82, 39, 91])

# Label each score as Pass or Fail using 50 as the cutoff
results = np.where(scores >= 50, "Pass", "Fail")
print(results)
```

```python
# Example: give a 5-point bonus to scores below 60
scores = np.array([45, 67, 52, 39, 91])
adjusted_scores = np.where(scores < 60, scores + 5, scores)

print("Original:", scores)
print("Adjusted:", adjusted_scores)
```

## 6. Aggregate and statistical functions

Aggregate functions summarize many values into one or more results.


```python
data = np.array([10, 20, 30, 40, 50])

print("Sum:", np.sum(data))
print("Mean:", np.mean(data))
print("Median:", np.median(data))
print("Minimum:", np.min(data))
print("Maximum:", np.max(data))
print("Standard deviation:", np.std(data))
print("Variance:", np.var(data))
print("25th percentile:", np.percentile(data, 25))
print("Cumulative sum:", np.cumsum(data))
```

### `argmin()` and `argmax()`

These return the index of a minimum or maximum value, rather than the value itself.


```python
data = np.array([40, 15, 90, 25])

print("Minimum value:", np.min(data))
print("Index of minimum:", np.argmin(data))
print("Maximum value:", np.max(data))
print("Index of maximum:", np.argmax(data))
```

## 7. Understanding `axis` in 2D arrays

For a 2D array:
- `axis=0` means operate down the rows, producing a result for each column.
- `axis=1` means operate across the columns, producing a result for each row.
- Omitting `axis` usually summarizes all elements.


```python
marks = np.array([
    [80, 70, 90],  # Student 1: three subjects
    [60, 85, 75],  # Student 2
    [95, 90, 85]   # Student 3
])

print("Marks:\n", marks)

# Sum of every element in the full array
print("Total of all marks:", np.sum(marks))

# Sum down rows: total for each subject (one result per column)
print("Column totals (axis=0):", np.sum(marks, axis=0))

# Sum across columns: total for each student (one result per row)
print("Row totals (axis=1):", np.sum(marks, axis=1))

# Mean score for each subject
print("Subject means:", np.mean(marks, axis=0))

# Mean score for each student
print("Student means:", np.mean(marks, axis=1))
```

**Teaching tip:** Always draw the rows and columns before explaining `axis`. Ask students whether they want one result per student (row) or one result per subject (column).


## 8. Sorting and unique values

`np.sort()` returns a sorted array. By default, it sorts along the last axis. For a 1D array, this means sorting the values from low to high.


```python
values = np.array([40, 10, 30, 10, 20])

print("Original:", values)
print("Sorted:", np.sort(values))
print("Original unchanged:", values)
print("Sorted descending:", np.sort(values)[::-1])
print("Unique values:", np.unique(values))
print("Unique values and counts:", np.unique(values, return_counts=True))
```

### Sort along an axis


```python
matrix = np.array([
    [3, 1, 2],
    [9, 7, 8]
])

print("Original:\n", matrix)
print("Sort each row (axis=1):\n", np.sort(matrix, axis=1))
print("Sort each column (axis=0):\n", np.sort(matrix, axis=0))
```

## 9. Basic random number generation

Use `np.random.default_rng()` to create a random-number generator. Providing a seed makes results reproducible.


```python
rng = np.random.default_rng(seed=42)

print("Random decimals:", rng.random(5))
print("Random integers from 1 to 10:", rng.integers(1, 11, size=5))
print("Random sample from choices:", rng.choice([10, 20, 30, 40], size=3))
```

The upper bound in `rng.integers(low, high)` is excluded. So `rng.integers(1, 11)` can generate integers from 1 through 10.
