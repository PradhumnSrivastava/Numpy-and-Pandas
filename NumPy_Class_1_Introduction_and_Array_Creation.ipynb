## 1. What is NumPy?

**NumPy** stands for **Numerical Python**. It is a Python library used for efficient numerical computation, especially when working with arrays, vectors, matrices, and large collections of numbers.

NumPy is useful in Data Science because it provides:
- Efficient multidimensional arrays.
- Mathematical operations on whole arrays.
- Statistical and linear-algebra functions.
- A foundation used by libraries such as Pandas and scikit-learn.

> Teaching note: NumPy is not a replacement for every Python list use case. It is especially useful when working with numerical data and performing repeated calculations.

## 3. Python list vs NumPy array

A Python list is a general-purpose container. A NumPy array is designed for efficient numerical operations.

### Example A: Python list


```python
numbers_list = [10, 20, 30, 40]

# Multiplying a list by 2 repeats the list.
print(numbers_list * 2)
```

### Example B: NumPy array

```python
numbers_array = np.array([10, 20, 30, 40])

# Multiplying an array by 2 multiplies every element.
print(numbers_array * 2)
```

### Key difference

| Operation | Python list | NumPy array |
|---|---|---|
| `values * 2` | Repeats the list | Multiplies each element by 2 |
| Numerical calculations | Often need loops/comprehensions | Many operations work element by element |
| Dimensions | Nested lists can represent tables | Native multidimensional array support |

Try these examples and explain the output in your own words.


```python
# Practice with both data structures
marks_list = [50, 60, 70]
marks_array = np.array([50, 60, 70])

print("List + list:", marks_list + marks_list)
print("Array + array:", marks_array + marks_array)
```

## 4. Creating a NumPy array with `np.array()`

The most common starting point is converting a Python list into a NumPy array.


```python
# One-dimensional (1D) array
arr_1d = np.array([10, 20, 30, 40, 50])
print(arr_1d)
```

```python
# Two-dimensional (2D) array: rows and columns
arr_2d = np.array([
    [10, 20, 30],
    [40, 50, 60]
])
print(arr_2d)
```

### Understand the dimensions

- **1D array:** A single sequence of elements.
- **2D array:** A table-like structure with rows and columns.
- A 3D array adds another dimension; we will study dimensions more deeply in a later class.


```python
# A simple 3D array example
arr_3d = np.array([
    [[1, 2], [3, 4]],
    [[5, 6], [7, 8]]
])
print(arr_3d)
```

## 5. Basic array creation functions

### `np.zeros()`
Creates an array filled with zeros. The shape is provided as an argument.


```python
zeros_1d = np.zeros(5)
zeros_2d = np.zeros((2, 3))

print("1D zeros:", zeros_1d)
print("2D zeros:\n", zeros_2d)
```

### `np.ones()`
Creates an array filled with ones.


```python
ones_1d = np.ones(4)
ones_2d = np.ones((3, 2))

print("1D ones:", ones_1d)
print("2D ones:\n", ones_2d)
```

### `np.full()`
Creates an array filled with a value you choose.


```python
filled_array = np.full((2, 3), 7)
print(filled_array)
```

### `np.arange()`
Creates values using a start, stop, and optional step. The **stop value is excluded**.


```python
print(np.arange(5))        # 0, 1, 2, 3, 4
print(np.arange(2, 10))     # 2 through 9
print(np.arange(2, 11, 2))  # step of 2
```

### `np.linspace()`
Creates a chosen number of evenly spaced values between a start and stop value. By default, both endpoints are included.


```python
print(np.linspace(0, 10, 5))
# Five evenly spaced values from 0 to 10, including both endpoints.
```

### `arange()` vs `linspace()`

- Use `arange(start, stop, step)` when you want to specify the step size.
- Use `linspace(start, stop, num)` when you want a specific number of evenly spaced values.


```python
# Compare the results
print("arange:", np.arange(0, 10, 2))
print("linspace:", np.linspace(0, 10, 5))
```

## 6. Creating arrays with different data types

NumPy arrays usually store elements using a common data type. You can inspect it with `.dtype` and explicitly request a type when creating an array.


```python
integer_array = np.array([1, 2, 3])
float_array = np.array([1.5, 2.5, 3.5])
float_from_list = np.array([1, 2, 3], dtype=float)

print(integer_array, "| dtype:", integer_array.dtype)
print(float_array, "| dtype:", float_array.dtype)
print(float_from_list, "| dtype:", float_from_list.dtype)
```
